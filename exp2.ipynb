```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("ML_csv.csv")   # Replace with your actual CSV file name

# Set plot style
sns.set_style("whitegrid")

# Numerical columns
num_cols = [
    'CGPA',
    'AttendancePercent',
    'AptitudeTestScore',
    'CodingTestScore',
    'SoftSkillsRating',
    'Salary Package'
]

# Missing Values Heatmap
plt.figure(figsize=(10, 6))
sns.heatmap(
    df.isnull(),
    cbar=False,
    cmap="viridis",
    yticklabels=False
)

plt.title("Missing Values Heatmap - Placement Prediction Dataset")
plt.xlabel("Columns")
plt.ylabel("Records")
plt.tight_layout()
plt.show()
```
